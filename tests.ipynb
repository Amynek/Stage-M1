```python
import numpy as np
from   numpy.polynomial.legendre import legval
from   scipy.special             import eval_legendre
```

```python
Ar = np.genfromtxt("Data/HCNAr_chinois.csv", delimiter=";", names=True, dtype=None, encoding="utf-8")

# Suppression des 8 dernières lignes
Ar = Ar[-83:-8]
```

```python
Theta = [0,10,20,30,40,50,60]
b = [1,2,3,4,5,6]
cosTh = np.cos(np.deg2rad(np.asarray(Theta, dtype=float)))
Theta, b, cosTh
```

```python
B=legval(cosTh,b)
B
```

```python
R=5
i = np.arange(6)
g0 = np.asarray([1,2,3,4,5,6], dtype=float)
g1 = np.asarray([7,8,9,1,2,3], dtype=float)
g2 = np.asarray([4,5,6,7,8,9], dtype=float)
g3 = np.asarray([1,2,3,4,5,6], dtype=float)
g = g0[i, None] + R * (g1[i, None] + R * (g2[i, None] + R *  g3[i, None]))
g
```

```python
Pl = np.asarray([eval_legendre(l, cosTh) for l in range(6)])
Pl
```

```python
np.shape(g * Pl)
```

```python
np.shape(np.sum(g * Pl, axis=0) * np.exp(B))
```

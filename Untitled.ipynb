```python
# 1. Jeu de paramètres aléatoire reproductible
rng = np.random.default_rng(42)
p_test = rng.uniform(-1, 1, size=40)

# 2. Quelques points R, Theta de test
R_test     = np.array([2.0, 3.0, 5.0, 8.0, 12.0])
Theta_test = np.array([0.0, 45.0, 90.0, 135.0, 180.0])

# 3. Résultat de référence (ancienne fonction)
ref = V_opt(params_He, R_test, Theta_test)

# 4. Résultat vectorisé — on passe p en (40, 1) et on squeeze
params_optd = params_He[:, None]                          # (40, 1)
new  = V_opt_vec(params_optd, R_test, Theta_test)[0]   # (1, M) → (M,)

# 5. Comparaison
print("Référence  :", ref)
print("Vectorisée :", new)
print("Différence max :", np.max(np.abs(ref - new)))
print("Quasi-identiques :", np.allclose(ref, new, rtol=1e-10, atol=1e-12))
```

```python
# 6. Test sur N jeux de paramètres simultanés
N = 50
P = rng.uniform(-1, 1, size=(40, N))   # (40, N)

# Référence : boucle sur chaque jeu
ref_batch = np.array([V_opt(P[:, k], R_test, Theta_test) for k in range(N)])  # (N, M)

# Vectorisée en une passe
new_batch = V_opt_vec(P, R_test, Theta_test)   # (N, M)

print("Erreur max sur le batch :", np.max(np.abs(ref_batch - new_batch)))
print("Batch OK :", np.allclose(ref_batch, new_batch, rtol=1e-10, atol=1e-12))
```

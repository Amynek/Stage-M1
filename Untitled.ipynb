# Import des données

```python
# Tableau : Ar-HCN-Chinois
Ar = np.genfromtxt("Data/HCNAr_chinois.csv", delimiter=";", names=True, dtype=None, encoding="utf-8")

## Suppression des 2 dernières lignes
#Ar = Ar[:-2]

# Supression des valeurs qui ne sont pas entre 2.5 et 10 Angstrom
Ar = Ar[-83:-8]

# Transformation du jeu de données pour avoir |R|Theta|E|
R_vals = Ar[Ar.dtype.names[0]]
theta_vals = np.array([float(name) for name in Ar.dtype.names[1:]])

R_col = np.repeat(R_vals, len(theta_vals))
Theta_col = np.tile(theta_vals, len(R_vals))

E_col = np.concatenate([Ar[name] for name in Ar.dtype.names[1:]]).reshape(len(theta_vals), len(R_vals)).T.ravel()

Ar_data = np.column_stack((R_col, Theta_col, E_col))

# Récupération des colonnes
R_ang = Ar_data[:, 0]
Theta_deg_inv = Ar_data[:, 1]
E_cmm1 = Ar_data[:, 2]

# Conversion des données
R_a0 = R_ang * 1.889726125
Theta_deg = 180.0 - Theta_deg_inv
E_mEh = E_cmm1 * 1000.0 / 219474.6313705
```

```python
R = torch.tensor(R_a0, dtype=torch.float64, device=device)
Theta = torch.tensor(Theta_deg, dtype=torch.float64, device=device)
E = torch.tensor(E_mEh, dtype=torch.float64, device=device)
```

```python
p0 = np.array([1.483407536976,  0.112015542111, -0.032843612191,  0.094365726134,  0.033820767387, -0.004629355901, 
              -135783.5865435, -325692.4344777, -57995.31109569, -125784.8469770,
               14.37087301922,  1.210291512784,  1.127352935061,  0.466341968270,  0.263961941831,  0.043692710808,
               0.044277336958, -0.150928579110,  0.029017675752, -0.029401214750,  0.013423710578,  0.009201449094, 
               0.008911721077,  0.050508135234, -0.008855531861,  0.013531186127, -0.008965289979, -0.005064185661,
              -0.002917801252, -0.005428112556,  0.001146212702, -0.001178096509,  0.001834942387,  0.000752982787, 
               0.000137627196,  0.000192853374, -0.000053626009, -0.000008219267, -0.000113691774, -0.000031076263 ])
p0 = np.load("params_opt_cf_lm_sigma.npy")
```

```python
p = torch.nn.Parameter(torch.tensor(p0, dtype=torch.float64, device=device))
```

```python
def loss_fn():

    pred = V_torch(p, R, Theta)

    eps = 1e-8

    abs_term = (pred - E)**2
    rel_term = abs_term / (E**2 + eps)

    alpha = 0.5  # à ajuster

    return torch.mean(alpha * rel_term + (1 - alpha) * abs_term)
```

```python
optimizer = torch.optim.Adam([p], lr=1e-3)

for i in range(2000):

    optimizer.zero_grad()

    loss = loss_fn()

    loss.backward()

    optimizer.step()

    if i % 100 == 0:
        print(f"Adam iter {i}, loss = {loss.item():.6e}")
```

```python
optimizer = torch.optim.LBFGS(
    [p],
    max_iter=500,
    line_search_fn="strong_wolfe"
)

def closure():
    optimizer.zero_grad()
    loss = loss_fn()
    loss.backward()
    return loss

for i in range(10):
    loss = optimizer.step(closure)
    print(f"LBFGS iter {i}, loss = {loss.item():.6e}")

p_opt = p.detach().cpu().numpy()
```
